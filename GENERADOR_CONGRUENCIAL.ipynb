```python
def generador(a, b, m, x0):
    x1 = (a*x0 + b)%m
    return x1
```

```python
a = 7
b = 5
m = 32
x0 = 0
n = 10
muestra = []
for i in range(n):
  nuevo = generador(a, b, m,x0=nuevo/m)
  muestra.append(nuevo)
print(muestra)
```
